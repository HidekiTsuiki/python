```python
# 練習問題20
def prime(x):
    if x < 2:          # 0 と 1 は素数ではない
        return False
    for i in range(2,x):
        if x % i == 0:
            return False
    return True

print(prime(13), prime(1), prime(91))
```

```python
# 練習問題20 （√x まで調べれば十分）
# x = a * b (a <= b) と分解できれば，a * a <= x なので，a <= √x である。
# よって，√x 以下の数で割り切れなければ，素数である。
def prime(x):
    if x < 2:
        return False
    i = 2
    while i * i <= x:
        if x % i == 0:
            return False
        i = i + 1
    return True

print(prime(13), prime(1), prime(91), prime(1000003))
```

```python
# 練習問題30
def numprime(x):
    c = 0
    for i in range(2,x+1):
        if prime(i):
            c += 1
    return c

numprime(10)
```

```python
# 練習問題40
import random
def pi():
    repeatnum = 1000000
    c = 0
    for _ in range(repeatnum):
        x = random.random()   # 0 以上 1 未満の乱数
        y = random.random()
        if x**2 + y**2 < 1:
            c = c + 1
    return 4 * c / repeatnum

pi()
```

```python
# 練習問題（ライプニッツの公式を if 文で）
def leibniz(n):
    x = 0
    for i in range(n):
        if i % 2 == 0:
            x = x + 1/(2*i+1)
        else:
            x = x - 1/(2*i+1)
    return x*4

leibniz(10000)
```

**while 文のバグ**

このプログラムは，i = 0 から始めて，x に i を足してから i を 1 増やし，i < 100 の間繰り返します。よって，x に足されるのは 0, 1, ..., 99 で，結果は 1 から 99 までの和 4950 になります。100 が足されていません。

直し方はいろいろあります。例えば，条件を `i <= 100` にする，i = 1 から始めて条件を `i <= 100` にする，i を 1 増やしてから x に足す，などです。

```python
i = 1
x = 0
while i <= 100:
    x = x + i
    i = i + 1
print(x)
```

```python
#練習問題50
def nextprime(n):
    while True:
        n += 1
        if prime(n):
            return n

nextprime(101)
```

```python
# 練習問題（コラッツ予想）
# n/2 だと float になってしまうので，整数の割り算 // を用いる。
def f(n):
    if n % 2 == 1:
        return 3*n + 1
    else:
        return n // 2

# 1. n, f(n), f(f(n)), ... を 1 になるまで表示する
def collatz(n):
    print(n, end=" ")
    while n != 1:
        n = f(n)
        print(n, end=" ")
    print()

# 2. n = 27 で確かめる
collatz(27)
```

```python
# 3. 1 になるまでの長さ（f を適用した回数）と，途中の最大値
def collatz_len(n):
    c = 0
    while n != 1:
        n = f(n)
        c += 1
    return c

def collatz_max(n):
    m = n
    while n != 1:
        n = f(n)
        if n > m:
            m = n
    return m

print(collatz_len(27), collatz_max(27))   # 111 回，最大値 9232
```

```python
# 4. 今まで現れた最長の長さ以上の長さが得られた時に，n とその長さを表示する
longest = 0
for n in range(1, 100000):
    c = collatz_len(n)
    if c >= longest:
        longest = c
        print(n, c)
```

```python
# 練習問題（n x n までの掛け算の表を表示する関数）
def table(n):
    for i in range(1, n+1):
        for j in range(1, n+1):
            print(f"{i*j:5}", end="")
        print()

table(12)
```
