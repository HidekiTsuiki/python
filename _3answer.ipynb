**while 文のバグ**

このプログラムは，i = 0 から始めて，x に i を足してから i を 1 増やし，i < 100 の間繰り返します。よって，x に足されるのは 0, 1, ..., 99 で，結果は 1 から 99 までの和 4950 になります。100 が足されていません。

直し方はいろいろあります。例えば，条件を `i <= 100` にする，i = 1 から始めて条件を `i <= 100` にする，i を 1 増やしてから x に足す，などです。

```python
i = 1
x = 0
while i <= 100:
    x = x + i
    i = i + 1
print(x)
```

```python
# 第2回の練習問題42 で作った prime（√x まで調べる版）
def prime(x):
    if x < 2:
        return False
    i = 2
    while i * i <= x:
        if x % i == 0:
            return False
        i = i + 1
    return True
```

```python
#練習問題50
def nextprime(n):
    while True:
        n += 1
        if prime(n):
            return n

nextprime(101)
```

```python
# 練習問題（コラッツ予想）
# n/2 だと float になってしまうので，整数の割り算 // を用いる。
def f(n):
    if n % 2 == 1:
        return 3*n + 1
    else:
        return n // 2

# 1. n, f(n), f(f(n)), ... を 1 になるまで表示する
def collatz(n):
    print(n, end=" ")
    while n != 1:
        n = f(n)
        print(n, end=" ")
    print()

# 2. n = 27 で確かめる
collatz(27)
```

```python
# 3. 1 になるまでの長さ（f を適用した回数）と，途中の最大値
def collatz_len(n):
    c = 0
    while n != 1:
        n = f(n)
        c += 1
    return c

def collatz_max(n):
    m = n
    while n != 1:
        n = f(n)
        if n > m:
            m = n
    return m

print(collatz_len(27), collatz_max(27))   # 111 回，最大値 9232
```

```python
# 4. 今まで現れた最長の長さ以上の長さが得られた時に，n とその長さを表示する
longest = 0
for n in range(1, 100000):
    c = collatz_len(n)
    if c >= longest:
        longest = c
        print(n, c)
```

```python
# 練習問題1
def mymax(u):
    m = u[0]
    for a in u:
        if a > m:
            m = a
    return m

def prod(u):
    p = 1
    for a in u:
        p = p * a
    return p

def count(u, a):
    num = 0
    for b in u:
        if b == a:
            num += 1
    return num

def index(u, a):
    """ a が u に含まれなければ，None を返す """
    for n in range(len(u)):
        if u[n] == a:
            return n

def ezero(u):
    for n in range(len(u)):
        if u[n] % 2 == 0:
            u[n] = 0

print(
    mymax ([5,3,2,1,3]),
    prod  ([5,3,2,1,3]),
    count ([5,3,2,1,3],3),
    index ([5,3,2,1,3],1)
)

u = [5,3,2,1,3]
ezero(u)
print(u)
```

```python
# 練習問題10
def triple(u):
    ret = []
    for a in u:
        ret.append(a*a*a)
    return ret

def more3(u):
    ret = []
    for a in u:
        if a >= 3:
            ret.append(a)
    return ret

def intersection(u,v):
    ret = []
    for a in u:
        if a in v:
            ret.append(a)
    return ret

def difference(u,v):
    ret = []
    for a in u:
        if a not in v:
            ret.append(a)
    return ret

def reverse(u):
    ret = []
    for a in u:
        ret.insert(0,a)
    return ret

# 次のような書き方も考えられます。
def reverse1(u):
    n = len(u)
    ret = []
    for i in range(n):
        ret.append(u[n-1-i])
    return ret

# ''' と ''' で囲んでも文字列が作れます。
#''' を用いると文字列に改行を含むことができます。
# それと、f による format を組み合わせて書いています。

print(f'''
triple の結果:{triple([1,2,3,4,5])}
more3 の結果:{more3([1,2,3,4,5])}
intersectionの結果:{intersection([2,4,6,8,10],[1,2,3,4,5])}
difference の結果: {difference([2,4,6,8,10], [1,2,3,4,5])}
reverse の結果: {reverse(range(5))}
reverse1 の結果: {reverse1(range(5))}
'''
)
```

## 練習問題

```python
# 練習問題（曜日）
# 2026 年 1 月 1 日は木曜日。日曜日を 0 とすると，木曜日は 4。
mon = [0, 31, 28, 31, 30, 31, 30, 31, 31, 30, 31, 30, 31]

def wday(m, n):
    """ 2026 年 m 月 n 日の曜日（日曜日が 0，…，土曜日が 6）"""
    days = sum(mon[1:m]) + (n - 1)   # 1 月 1 日から何日後か
    return (4 + days) % 7

week = ["日曜日", "月曜日", "火曜日", "水曜日", "木曜日", "金曜日", "土曜日"]

def wdaystr(m, n):
    return week[wday(m, n)]

print(wdaystr(1, 1), wdaystr(10, 6), wdaystr(12, 25))
```

```python
# 練習問題60
def divs(x):
    """ x の約数のリスト """
    ret = []
    for i in range(1, x+1):
        if x % i == 0:
            ret.append(i)
    return ret

d = []
for x in range(1, 101):
    d.append(divs(x))
print(d[:12])
```

```python
# 練習問題70
def complete(x):
    return sum(divs(x)) - x == x    # x 自身を除いた約数の和

for x in range(1, 10000):
    if complete(x):
        print(x)
```

```python
# 練習問題（wday の拡張）
# 西暦 2001 年 1 月 1 日は月曜日（1）。2001 年以降の日付を扱う。
def uruu(y):
    return ((y % 4 == 0) and (y % 100 != 0)) or (y % 400 == 0)

def wday(y, m, n):
    days = 0
    for yy in range(2001, y):
        if uruu(yy):
            days += 366
        else:
            days += 365
    days += sum(mon[1:m]) + (n - 1)
    if m > 2 and uruu(y):
        days += 1
    return (1 + days) % 7

print(week[wday(2026, 1, 1)], week[wday(2024, 3, 1)], week[wday(2100, 3, 1)])
```

```python
# 練習問題8
# カレンダーは 400 年で元に戻る（400 年の日数 146097 は 7 で割り切れる）ので，
# 400 年分の各月の 13 日の曜日を数えればよい。
count13 = [0]*7
for y in range(2001, 2401):
    for m in range(1, 13):
        count13[wday(y, m, 13)] += 1

for i in range(7):
    print(week[i], count13[i])
# 金曜日が 688 回で，一番多い。
```

```python
# 練習問題9
# 表を 1, 裏を -1 で表現する。ひっくり返すのは -1 を掛けるだけで済む。
coin = [1]*1001      # coin[1] から coin[1000] を使う
for k in range(2, 1001):
    for j in range(k, 1001, k):
        coin[j] = coin[j] * (-1)

for i in range(1, 1001):
    if coin[i] == 1:
        print(i, end=" ")
print()
# i 番目のコインは，i の 1 以外の約数の個数だけ裏返される。
# 約数の個数が奇数になるのは i が平方数の時だけなので，表になっているのは平方数番目のコイン。
```
